`````

```{figure} pict/1_word/besmele.png
---
scale: 40%
align: center
---
```

```{admonition} Quran
:class: toggle

- And ˹remember˺ when Moses prayed for water for his people, We said, “Strike the rock with your staff.” Then twelve springs gushed out, ˹and˺ each tribe knew its drinking place. ˹We then said,˺ “Eat and drink of Allah’s provisions, and do not go about spreading corruption in the land.” (Al-Baqarah 2:60)

- We ordered, “O fire! Be cool and safe for Abraham!” (Al-Anbiya 21:69)

- Do not eat of what is not slaughtered in Allah’s Name. For that would certainly be ˹an act of˺ disobedience. Surely the devils whisper to their ˹human˺ associates to argue with you. If you were to obey them, then you ˹too˺ would be polytheists.(Al-An'am 6:121)

- O believers! Do not waste your charity with reminders ˹of your generosity˺ or hurtful words, like those who donate their wealth just to show off and do not believe in Allah or the Last Day. Their example is that of a hard barren rock covered with a thin layer of soil hit by a strong rain—leaving it just a bare stone. Such people are unable to preserve the reward of their charity. Allah does not guide ˹such˺ disbelieving people. (Al-Baqarah 2:264)

## Warm Up Questions
1. What does "Bismillah" mean to you, and how does saying it influence your daily actions?


```{admonition} Remark
:class: toggle
Bismillah means "In the Name of God." Saying it reminds me to begin my actions with good intentions, seek God's help, and remember His blessings.

2. What is the difference between being humble and being proud, especially when facing difficulties?

```{admonition} Remark
:class: toggle
A humble person recognizes their limitations and seeks help, while a proud person relies only on themselves. Humility brings strength and confidence through trust in God.


## Background
```{figure} pict/1_word/background.png
---
scale: 30%
align: left
---
```
In The First Word from The Words, Bediüzzaman Said Nursi explains the spiritual meaning of Bismillah, “In the Name of God.” He begins with a story of two travellers crossing a desert: one humbly travels under the protection of a tribal chief, while the other relies on himself and suffers hardships.

The story becomes a comparison for human life. Nursi describes people as dependent on God’s power and mercy and points to seeds, trees, gardens, animals, roots, and leaves as signs of Divine generosity. He concludes that blessings call for remembrance, thanks, and reflection.


**Ar-Rahman and Ar-Rahim: Allah’s Mercy:** 
Ar-Rahman (الرَّحْمَٰن) and Ar-Rahim (الرَّحِيم) are two beautiful names of Allah that express His mercy and compassion. Although both names come from the Arabic word rahmah, meaning mercy, they emphasize different aspects of Allah’s mercy. Ar-Rahman means “The Most Merciful” and refers to Allah’s vast and unlimited mercy that embraces all creation, including believers and nonbelievers. For example, Allah provides sunlight, water, food, and countless blessings to everyone. Ar-Rahim means “The Especially Merciful” and emphasizes Allah’s continuous mercy, particularly His special mercy toward believers through guidance, forgiveness, and eternal blessings. When we say Bismillah ir-Rahman ir-Rahim, we remember that Allah’s mercy is both universal and deeply personal, and that every blessing we receive comes from His kindness and generosity.

## Main Passage

```{figure} pict/1_word/main_passage.png
---
scale: 30%
align: right
---
```

[Download Main Passage](1_word.pdf)


<font color="red">Bismillah</font>, “In the Name of God,” is the start of all things good. We too shall start with it. Know, O my soul! Just as this blessed phrase is a mark of Islam, so too it is constantly recited by all beings through their tongues of <font color="red">disposition</font>. If you want to know what an <font color="red">inexhaustible</font> strength, what an unending source of <font color="red">bounty</font> is <font color="red">Bismillah</font>, listen to the following story which is in the form of a comparison.

It goes like this: Someone who makes a journey through the deserts of Arabia has to travel in the name of a tribal chief and enter under his protection, for in this way he may be saved from the assaults of bandits and secure his needs. On his own he will perish in the face of innumerable enemies and needs.

And so, two men went on such a journey and entered the desert. One of them was modest and humble, the other proud and <font color="red">conceited</font>. The humble man assumed the name of a tribal chief, while the proud man did not. The first travelled safely wherever he went. If he encountered bandits, he said: “I am travelling in the name of such-and-such tribal leader,” and they did not <font color="red">molest</font> him. If he enters a tent, he was treated respectfully due to the name.

But the proud man suffered indescribable <font color="red">calamities</font> throughout his journey. He both <font color="red">trembled</font> before everything and begged from everything. He was abased and became an object of <font color="red">scorn</font>.

My proud soul! You are the traveller, and this world is a desert. Your <font color="red">impotence</font> and <font color="red">poverty</font> have no limit, and your enemies and needs are endless. Since it is thus, take the name of the <font color="red">Pre-Eternal</font> Ruler and <font color="red">Post-Eternal</font> Lord of the desert and be saved from begging before the whole universe and trembling before every event.

Yes, this phrase is a <font color="red">treasury</font> so blessed that your infinite <font color="red">impotence</font> and <font color="red">poverty</font> bind you to an infinite power and <font color="red">mercy</font>; it makes your <font color="red">impotence</font> and <font color="red">poverty</font> a most acceptable <font color="red">intercessor</font> at the Court of One All-Powerful and <font color="red">Compassionate</font>.

The person who acts saying, “In the Name of God,” resembles someone who enrolls in the army. He acts in the name of the government; he has fear of no one; he speaks, performs every matter, and withstands everything in the name of the law and the name of the government.

At the beginning we said that all beings say “In the Name of God” through the tongue of <font color="red">disposition</font>. Is that so? Indeed, it is so.

If you were to see that a single person had come and had driven all the inhabitants of a town to a place by force and <font color="red">compelled</font> them to work, you would be certain that he had not acted in his own name and through his own power, but was a soldier, acting in the name of the government and relying on the power of the king.

In the same way, all things act in the name of Almighty God, for minute things like seeds and grains bear huge trees on their heads; they raise loads like mountains.

That means all trees say: “In the Name of God,” fill their hands from the <font color="red">treasury</font> of <font color="red">mercy</font>, and offer them to us.

All gardens say: “In the Name of God,” and become <font color="red">cauldrons</font> from the kitchens of Divine power in which are cooked numerous varieties of different foods.

All blessed animals like cows, camels, sheep, and goats, say: “In the Name of God,” and produce springs of milk from the <font color="red">abundance</font> of <font color="red">mercy</font>, offering us a most delicate and pure food like the water of life in the name of the Provider.

The roots and <font color="red">rootlets</font>, soft as silk, of plants, trees, and grasses say: “In the Name of God,” and pierce and pass through hard rock and earth. Mentioning the name of God, the name of the Most Merciful, everything becomes <font color="red">subjected</font> to them.

The roots spreading through hard rock and earth and producing fruits as easily as the branches spread through the air and produce fruits, and the delicate green leaves retaining their moisture for months in the face of extreme heat, deal a slap in the mouths of <font color="red">Naturalists</font> and jab a finger in their blind eyes, saying: “Even heat and hardness, in which you most trust, are under a command. For like the Staff of Moses, each of those silken <font color="red">rootlets</font> conform to the command of, And We said, O Moses, strike the rock with your staff, and split the rock. (Qur’an, 2:60) And the delicate leaves fine as cigarette paper recite the verse, O fire be coolness and peace against the heat of the fire, each like the members of Abraham (UWP: Upon Whom Be Peace). (Qur’an, 21:69)

Since all things say: “In the Name of God,” and bearing God’s bounties in God’s name, give them to us, we too should say: “In the Name of God.” We should give in the name of God, and take in the name of God. And we should not take from <font color="red">heedless</font> people who neglect to give in God’s name.

Question: We give a price to people, who are like tray-bearers. So what price does God want, Who is the true owner?

The Answer: Yes, the price the <font color="red">True Bestower of Bounties</font> wants in return for those valuable bounties and goods is three things: one is <font color="red">remembrance</font>, another is thanks, and the other is <font color="red">reflection</font>. Saying, “In the Name of God” at the start is <font color="red">remembrance</font>, and, “All praise be to God” at the end is thanks. And perceiving and thinking of those bounties, which are priceless wonders of art, being miracles of power of the Unique and Eternally <font color="red">Besought</font> One and gifts of His <font color="red">mercy</font>, is <font color="red">reflection</font>.

However foolish it is to kiss the foot of a lowly man who conveys to you the precious gift of a king and not to recognize the gift’s owner, it is a thousand times more foolish to praise and love the apparent <font color="red">bestower</font> of <font color="red">bounty</font> and forget the <font color="red">True Bestower of Bounties</font>.

O my soul! If you do not wish to be foolish in that way, give in God’s name, take in God’s name, begin in God’s name, and act in God’s name. And that’s the matter in a nutshell!



## Challenging Question 

1. The author describes Bismillah as an "inexhaustible strength" and an "unending source of bounty." How can a short phrase carry such great spiritual meaning?



```{admonition} Remark
:class: toggle
Bismillah connects a person to God's infinite power and mercy. Although the phrase is short, it reminds believers that their true strength and blessings come from God.

2. The author identifies remembrance, thanks, and reflection as three responses to God's blessings. How are these three concepts connected, and how can they transform our daily lives?

```{admonition} Remark
:class: toggle

Remembrance means beginning in God's name, thanks means expressing gratitude for His blessings, and reflection means recognizing His power and mercy in creation. Together, they help us live with greater awareness, gratitude, and responsibility.

## Summary 

```{figure} pict/1_word/summary.png
---
scale: 30%
align: left
---
```
Bediüzzaman teaches that saying Bismillah is more than speaking a phrase: it means beginning and acting with awareness of God. The humble traveller in the desert finds safety through a chief’s protection, while the proud traveller suffers by relying only on himself. In the same way, human beings can recognize their own limitations and turn toward God’s power and mercy.

Nature offers further examples. Small seeds grow into large trees, gardens produce food, and animals provide milk. The author sees these blessings as signs that creation acts through Divine power rather than independently.

Finally, he describes three responses to blessings: remembrance at the beginning, thanks at the end, and reflection on their source and meaning. The practical lesson is to give, receive, begin, and act in God’s name.

## Reading Comprehension Questions
```{figure} pict/rc.png
---
scale: 50%
align: right
---
```
1. What does “Bismillah” mean, and why does the author begin with it?
2. Why does a traveller in the Arabian desert seek the protection of a tribal chief?
3. How do the humble traveller and the proud traveller behave differently?
4. What does the desert represent when the author addresses his soul?
5. How is someone who says “In the Name of God” compared to a soldier?
6. What examples from trees, gardens, and animals illustrate the message?
7. What do the roots and leaves show about heat, hardness, and Divine command?
8. According to the passage, how should we give and receive blessings?
9. What three things does the True Bestower of Bounties ask in return?
10. What does the author mean by forgetting the true owner of a gift? 


## Critical Thinking Questions
```{figure} pict/ct.png
---
scale: 50%
align: left
---
```
1. How might beginning an ordinary task with Bismillah change your intention?
2. Why can humility give someone strength while pride can make them vulnerable?
3. What modern situations resemble the two travellers’ different approaches?
4. How does recognizing dependence on God affect anxiety or self-confidence?
5. What can a seed growing into a tree teach us about blessings we overlook?
6. How might you practice remembrance, thanks, and reflection during a meal?
7. Why is it easy to thank the person delivering a gift but forget its ultimate source?
8. How can gratitude influence the way we use natural resources?
9. What does the phrase “tongue of disposition” suggest about how nature communicates?
10. Choose one daily habit: how could you begin it, carry it out, and finish it in the spirit of the passage?


## Glossary
```{figure} pict/glossary.png
---
scale: 80%
align: right
---
```

| Word | Definition |
|------|------------|
| Abundance | A very large amount of something; more than enough. |
| Besought | Asked earnestly or begged for help; past tense of beseech. |
| Bestower | Someone who gives gifts, blessings, or benefits. |
| Bismillah | An Arabic phrase meaning "In the Name of God." |
| Bounty | A generous gift, blessing, or something given in large amounts. |
| Calamities | Serious disasters or events that cause suffering or damage. |
| Cauldrons | Large, deep pots used for cooking or boiling. |
| Compassionate | Showing kindness and concern for others who are suffering. |
| Compelled | Forced or strongly driven to do something. |
| Conceited | Too proud of yourself and your abilities. |
| Disposition | A person's usual character, attitude, or nature. |
| Heedless | Not paying attention or being careless about something important. |
| Impotence | A lack of power or ability to do something. |
| Inexhaustible | So great that it can never be used up. |
| Intercessor | Someone who speaks or asks for help on behalf of another person. |
| Mercy | Kindness, forgiveness, or compassion shown to others. |
| Molest | To disturb, bother, or harm someone. |
| Naturalists | People who study nature or explain things through natural causes. |
| Poverty | The condition of having very little or lacking what is needed. |
| Post-Eternal | Having no end; continuing forever into the future. |
| Pre-Eternal | Having no beginning; existing forever in the past. |
| Reflection | Deep and careful thought about something. |
| Remembrance | Keeping someone or something in your thoughts and memory. |
| Rootlets | Very small, thin roots growing from larger roots. |
| Scorn | A feeling of strong disrespect or contempt. |
| Subjected | Made to experience something or placed under someone's power or control. |
| Treasury | A place or source where valuable things are kept. |
| Trembled | Shook slightly because of fear, cold, or strong emotion. |
| True Bestower of Bounties | God, the One who gives all blessings and good things. |
