`````

```{figure} pict/13_word_2_station/???.png
---
scale: 40%
align: center
---

```

## Warm Up Questions
1. Why do you think youth is considered one of the most valuable periods of life?

```{admonition} Remark
:class: toggle
Youth is a time of energy, opportunities, and personal growth. It is valuable because the choices we make during this period can shape our future.

2. Can something that brings immediate pleasure cause harm in the long run? Give an example

```{admonition} Remark
:class: toggle
Yes. For example, spending too much time on entertainment may feel enjoyable but can lead to wasted time, neglected responsibilities, and regret.

3. How can belief in God influence the way we think about our future?

```{admonition} Remark
:class: toggle
Belief in God can give us hope, purpose, and peace of mind. It can help us face uncertainty and make choices that benefit both our present life and the hereafter.

## Background
```{figure} pict/13_word_2_station/background.png
---
scale: 40%
align: left
---
```

This reading is taken from The Thirteenth Word, Second Station, of The Words, part of the Risale-i Nur Collection written by Bediüzzaman Said Nursi (1877–1960).

The Thirteenth Word focuses on the guidance of the Qur'an and its ability to illuminate human life. In its First Station, Bediüzzaman compares the wisdom of the Qur'an with human philosophy, highlighting how Qur'anic teachings provide a meaningful understanding of existence and the universe. The Second Station applies this guidance to practical challenges in everyday life, particularly those faced by young people. It emphasizes the importance of making wise choices, recognizing the temporary nature of worldly life, and preparing for the hereafter.

The Second Station of the Thirteenth Word addresses the spiritual challenges faced by young people, particularly the temptations of worldly pleasures and the tendency to overlook the hereafter. In this section, titled “A Warning, Lesson and Reminder Given to a Number of Unhappy Youths,” Bediüzzaman responds to a group of young people seeking guidance on how to protect themselves from harmful desires and make better choices.

The passage emphasizes that youth is a temporary but precious gift that should be used wisely. Through powerful comparisons, including poisonous honey, a gallows, and a valuable lottery ticket, Bediüzzaman illustrates the difference between temporary pleasures and lasting happiness. He explains how belief, moral responsibility, and religious duties can give life deeper meaning and prepare individuals for eternal happiness.

As you read, consider how the author's message connects to the challenges young people face today and how faith can influence personal choices, happiness, and one's understanding of the future.

## Main Passage

```{figure} pict/13_word_2_station/main_passage.png
---
scale: 40%
align: right
---
```

[Download Main Passage](13_word_2_station.pdf)
**A Warning, Lesson and Reminder Given to a Number of Unhappy Youths**

One day a number of bright youths came to me, seeking an effective<font color="red"> deterren</font>t in order to guard themselves against the dangers arising from life, youth, and the lusts of the soul. I said to these youths as I had to those who had previously sought help from the Risale-i Nur: Your youth will definitely leave you, and if you do not remain within the bounds of the licit, it will be lost, and rather than its pleasures, it will bring you calamities and suffering in this world, in the grave, and in the hereafter. But if, through Islamic training, you spend the bounty of your youth as thanks honourably, in uprightness and obedience, it will in effect remain perpetually and will be the cause of gaining eternal youth. 

As for life, if it is without belief, or because of rebelliousness belief is ineffective, it will produce pains, sorrows and grief far exceeding the superficial, fleeting enjoyment it brings. Because, since, contrary to the animals, man possesses a mind and he thinks, he is connected to both the present time, and to the past and the future. He can obtain both pain and pleasure from them. Whereas, since the animals do not think, the sorrows arising from the past and the fears and anxieties arising from the future do not spoil their pleasure of the present. Especially if the pleasure is illicit; then it is like an altogether poisonous honey. 

That is to say, from the point of view of the pleasure of life, man falls to a level a hundred times lower than the animals. In fact, life for the people of misguidance and heedlessness, and indeed their existence, rather their world, is the day in which they find themselves. From the point of view of their misguidance, all the time and universes of the past are non-existent, are dead. So their intellects, which connect them to the past and the future, produce darkness, blackness for them. Due to their lack of belief, the future is also non-existent. Furthermore, because they think, the eternal separations resulting from this non-existence continuously produce darkness for their lives. 

But if belief gives life to life, through its light both the past and the future are illuminated and find existence. Like present time, it produces elevated, spiritual pleasures and lights of existence for the spirit and heart in respect of belief. There is an explanation of this truth in the ‘Seventh Hope’ in The Treatise for the Elderly. You may refer to that. 

Life is thus. If you want the pleasure and enjoyment of life, give life to your life through belief, and adorn it with religious duties, and preserve it by abstaining from sins. Concerning the fearsome reality of death, which is demonstrated by deaths every day, everywhere, at all times, I shall explain it to you with a comparison, in the same way that I told the other youths. 

For example, a gallows has been erected here in front of your eyes. Beside it is a lottery office, but one which gives tickets for truly huge prizes. We people here are ten people, whether we like it or not, we shall be summoned there; there is no other alternative. They will call us, and since the time is secret, any minute they may say either: “Come and collect the ticket for your execution! Mount the gallows!” Or: “A ticket to win a prize of millions of dollars’ worth of gold has come up for you. Come and collect it!” 

While waiting for them to say this, two people suddenly appear at the door. One of them is a scantily dressed woman, beautiful and deceiving. In her hand is some apparently extremely delicious, but in fact poisonous, candy, which she has brought wanting us to eat it. The other is an undeceiving and undeceivable serious person. He enters behind the woman and says: “I have brought you a talisman, a lesson. If you study it, and if you do not eat that candy, you will be saved from the gallows. With this talisman, you will receive your ticket for the matchless prize. 

Look, you see with your own eyes that those who eat the honey mount those gallows, and until that time they suffer dreadful stomach pains from the poison of the candy. And who it is that will receive the ticket for the large prize is not apparent; it seems that they too mount the gallows. But there are millions of witnesses who testify that they can enter the prize arena easily. 

So, look from the windows! The highest officials and the high-ranking persons concerned with this business proclaim with loud voices: ‘Just as you see with the clear certainty of your own eyes those mounting the gallows, so be certain as daylight, with no doubt or misgiving, that those with the talisman receive the ticket for the prize.

Thus, like the comparison, since the dissolute pleasures of youth in the sphere of the illicit, which are like poisonous honey, lose belief, which is the ticket for an eternal treasury and the passport for everlasting happiness, a person who indulges in them descends to death, which is like the gallows, and to the tribulations of the grave, which is like the door to eternal darkness. 

And since the appointed hour is unknown, its executioner, not differentiating between young and old, may come at any time to cut off your head. If you give up illicit desires, which are like the poisonous honey, and acquire belief and perform the religious duties, which are the Qur’anic talisman, one hundred and twenty-four thousand prophets (Peace be upon them) together with innumerable saints and people of truth have unanimously announced that you shall receive the ticket for the treasury of eternal happiness which comes up from the extraordinary lottery of human destiny. And they have pointed to traces of it. 

In Short: Youth will go….

May Almighty God save us and you from the alluring temptations of this time, and preserve us from them. Amen.



## Challenging Question 
1. Why does Bediüzzaman argue that human intelligence can become a source of suffering rather than happiness, and how does belief transform this experience?

```{admonition} Remark
:class: toggle

Unlike animals, human beings possess intelligence that connects them to the past and future. Without belief, memories of loss and fears about the future can overshadow present happiness. Bediüzzaman argues that belief transforms this experience by illuminating the past and future with spiritual meaning and hope. Thus, intelligence, which might otherwise increase suffering, becomes a means of understanding life more deeply and experiencing spiritual peace.


2. How does Bediüzzaman's comparison of poisonous honey, the gallows, and the talisman challenge the idea that immediate pleasure is the measure of true happiness?

```{admonition} Remark
:class: toggle

The comparison demonstrates that appearances can be deceptive. Poisonous honey symbolizes illicit pleasures that offer immediate enjoyment but lead to suffering. The gallows represent death and its frightening consequences, while the talisman symbolizes belief and religious duties that lead toward eternal happiness. Through these symbols, Bediüzzaman challenges readers to evaluate their choices not merely by the pleasure they provide in the present but by their lasting consequences. True happiness, according to the passage, comes from making choices that protect both one's spiritual well-being and eternal future.

## Summary 

```{figure} pict/13_word_2_station/summary.png
---
scale: 30%
align: left
---
```
Bediüzzaman begins by warning that youth will inevitably pass. He explains that young people should use this valuable period of life responsibly, observing religious principles and avoiding harmful temptations.

He then discusses the relationship between belief and happiness. Human beings possess minds that connect them to the past and future. Without belief, he argues, memories of the past and uncertainty about the future can become sources of suffering. With belief, however, these same dimensions of time can become sources of hope and spiritual comfort.

The author illustrates his message through an imaginative comparison. A group of people awaits an uncertain outcome involving either a terrible punishment or an extraordinary reward. Two figures appear: one offers attractive but poisonous candy, while the other offers a talisman that provides protection and the promise of a great prize.

Through this comparison, Bediüzzaman represents illicit pleasures as poisonous honey, death as the gallows, and belief and religious duties as the means of attaining eternal happiness.

He concludes by reminding readers that death can arrive at any age and that young people should not postpone their spiritual responsibilities.

<br><br><br><br><br><br><br><br>

## Main Ideas
Why does Bediüzzaman emphasize that youth will pass?
- He wants young people to recognize that youth is not permanent. Its energy, opportunities, and pleasures should be used wisely rather than wasted on activities that may lead to regret. In his view, youth becomes truly valuable when it is directed toward gratitude, uprightness, and obedience to God.

How does belief change a person's experience of life?
- According to the passage, human beings experience suffering because they remember the past and anticipate the future. Bediüzzaman argues that belief illuminates both, allowing people to understand life through a spiritual perspective rather than seeing it as limited to temporary experiences.

What does poisonous honey represent?
- Poisonous honey represents pleasures that appear sweet and attractive but eventually produce harmful consequences. The comparison emphasizes the difference between immediate enjoyment and long-term well-being.

What do the gallows and lottery symbolize?
- The gallows represent death and, within the comparison, the frightening consequences associated with spiritual loss. The lottery and its extraordinary prize symbolize the possibility of eternal happiness. The uncertainty surrounding the summons reminds readers that the time of death is unknown.

What is the meaning of the talisman?
- The talisman represents the spiritual guidance associated with belief and religious duties. In the passage, following this guidance is presented as the way to avoid spiritual danger and attain eternal happiness.




## Key Takeaways 
The central message of the passage is that youth is temporary, but the consequences of how it is spent can be lasting.

Bediüzzaman encourages young people to recognize the value of their time, distinguish between beneficial and harmful pleasures, strengthen their belief, and make choices with both their present lives and eternal future in mind.

## Reading Comprehension Questions
```{figure} pict/rc.png
---
scale: 50%
align: right
---
```
1. Why did the young people approach Bediüzzaman for advice?
2. What does the author say will inevitably happen to youth?
3. Under what conditions does the author say youth can become a source of eternal happiness?
4. Why does Bediüzzaman compare the experience of human beings with that of animals?
5. How can memories of the past and concerns about the future affect human happiness?
6. What does the author mean when he says that belief gives life to life?
7. What two possible outcomes are presented in the lottery comparison?
8. Who are the two figures who appear while the people are waiting?
9. Why is the candy described as attractive but poisonous?
10. What does the serious person offer, and what does it represent?
11. Why does Bediüzzaman emphasize that the time of death is unknown?
12. What advice does Bediüzzaman give at the end of the passage?


## Critical Thinking Questions
```{figure} pict/ct.png
---
scale: 50%
align: left
---
```

1. Why might young people find it difficult to appreciate the value of youth while they are experiencing it?
2. How can social media, entertainment, or peer pressure become examples of the poisonous honey described in the passage?
3. Is every enjoyable activity necessarily harmful? How does the passage distinguish between lawful enjoyment and illicit pleasure?
4. How might belief help someone deal with regret about the past or anxiety about the future?
5. Why does Bediüzzaman use a dramatic comparison involving death and extraordinary rewards instead of simply giving advice?
6. What is the difference between living only for the present and living with long-term goals?
7. How can young people balance enjoying their youth with fulfilling their moral and religious responsibilities?
8. What are some modern examples of choices that provide immediate satisfaction but may have negative long-term consequences?
9. How does awareness of death influence the way a person makes decisions?
10. If you were explaining the central message of this passage to a friend, what would you say?


## Glossary
```{figure} pict/glossary.png
---
scale: 80%
align: right
---
```

| Word | Definition |
|------|------------|
| **Deterrent** | Something that discourages a person from doing something harmful. |
| **Lusts of the soul** | Strong desires or temptations that may lead someone away from moral principles. |
| **Licit** | Permitted or lawful. |
| **Illicit** | Forbidden or unlawful. |
| **Uprightness** | Honesty, integrity, and morally correct behavior. |
| **Perpetually** | Continuously or forever. |
| **Fleeting** | Lasting for only a short time. |
| **Heedlessness** | Carelessness or lack of awareness about important responsibilities. |
| **Misguidance** | Following a mistaken or spiritually harmful path. |
| **Abstaining** | Deliberately avoiding or refraining from something. |
| **Gallows** | A structure historically used for execution by hanging. |
| **Talisman** | An object or symbol believed to provide protection; used metaphorically for spiritual guidance. |
| **Tribulations** | Serious difficulties, suffering, or trials. |
| **Dissolute** | Lacking moral restraint; given to excessive or improper pleasures. |
| **Appointed hour** | The predetermined time of death. |
| **Eternal treasury** | A metaphor for everlasting happiness in the hereafter. |
